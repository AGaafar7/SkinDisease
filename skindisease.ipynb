**Train Model**

```python
!pip install ultralytics
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=120,
    imgsz=640,
    batch=16,
    cos_lr=True,
    patience=30,
    close_mosaic=15,
    augment=True,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=5,
    translate=0.1,
    scale=0.5,
    fliplr=0.5
)
```

**Validate Model**

```python
from ultralytics import YOLO

best_model = YOLO("runs/detect/train/weights/best.pt")
metrics = best_model.val()
print(metrics)
```

**Export Model**

```python
best_model.export(
    format="ncnn",
    imgsz=640,
    simplify=True,
    opset=12,
    dynamic=False,
    nms=True,
    conf=0.25
)
```
